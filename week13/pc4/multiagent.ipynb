```python
import random
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# ── CONFIGURACIÓN ──────────────────────────────────────
GRID = 18

# Filas/cols 7 y 8 forman una AVENIDA de doble carril:
#   fila 7 → izquierda, fila 8 → derecha  (doble sentido)
#   col  7 → arriba,    col  8 → abajo
STREET_ROWS = [2, 7, 8, 14]
STREET_COLS = [2, 7, 8, 14]
SR = frozenset(STREET_ROWS)
SC = frozenset(STREET_COLS)
INTERSECTIONS = frozenset((r, c) for r in STREET_ROWS for c in STREET_COLS)

AVE_ROW_DIV = 8   # y donde se divide la avenida horizontal (entre fila 7 y 8)
AVE_COL_DIV = 8   # x donde se divide la avenida vertical  (entre col  7 y 8)

DIR_R=(0,1); DIR_L=(0,-1); DIR_D=(1,0); DIR_U=(-1,0)
ROW_DIR = {2:DIR_R, 7:DIR_L, 8:DIR_R, 14:DIR_L}
COL_DIR = {2:DIR_D, 7:DIR_U, 8:DIR_D, 14:DIR_U}

# Rangos de manzanas (celdas que NO son calles)
def _gaps(ss, total):
    out, s = [], None
    for i in range(total):
        if i in ss:
            if s is not None: out.append((s, i)); s = None
        else:
            if s is None: s = i
    if s is not None: out.append((s, total))
    return out

ROW_GAPS = _gaps(SR, GRID)   # [(0,2),(3,7),(9,14),(15,18)]
COL_GAPS = _gaps(SC, GRID)   # idem
BLOCKS   = [(r0,r1,c0,c1) for r0,r1 in ROW_GAPS for c0,c1 in COL_GAPS]

# Colores
CAR_PAL = ["#1A1A1A","#1C2128","#0D1B2A","#1B2631","#212529","#2C3E50","#17202A","#0A3D62"]
LRED="#FF3030"; LYEL="#FFD700"; LGRN="#00E64D"; LOFF="#151515"
C_BLK="#8FA888"; C_RD="#4A5568"; C_BG="#2D3748"; C_DIV="#FFD700"

PH_G=0; PH_Y=1; PH_R=2
G_T=12; Y_T=3; CYCLE=(G_T+Y_T)*2  # 30 ticks

SUBSTEPS=4; TICKS=100; NV=16

# 5 semáforos compartidos por carriles de la misma avenida
# (intersecciones menores calle×calle no tienen semáforo)
LIGHT_GROUPS = [
    ([(7,2),(8,2)],         0),
    ([(7,7),(7,8),(8,7),(8,8)], CYCLE//3),
    ([(7,14),(8,14)],       2*CYCLE//3),
    ([(2,7),(2,8)],         CYCLE//6),
    ([(14,7),(14,8)],       CYCLE//2),
]

# ── AGENTES ────────────────────────────────────────────
class TL:
    """Semáforo trifásico. Un objeto puede ser compartido por varios carriles."""
    def __init__(self, offset=0): self.t = offset % CYCLE
    def step(self): self.t = (self.t+1) % CYCLE
    def _ph(self):
        t = self.t
        if   t < G_T:        return PH_G, PH_R
        elif t < G_T+Y_T:    return PH_Y, PH_R
        elif t < G_T*2+Y_T:  return PH_R, PH_G
        else:                 return PH_R, PH_Y
    def hp(self): return self._ph()[0]
    def vp(self): return self._ph()[1]
    def ok_h(self): return self.hp()==PH_G
    def ok_v(self): return self.vp()==PH_G


class Car:
    def __init__(self, vid):
        self.vid=vid; self.color=random.choice(CAR_PAL)
        self.h=random.choice([True,False])
        if self.h:
            self.r=random.choice(STREET_ROWS); self.c=random.randint(0,GRID-1)
            self.d=ROW_DIR[self.r]
        else:
            self.c=random.choice(STREET_COLS); self.r=random.randint(0,GRID-1)
            self.d=COL_DIR[self.c]
        self.w=False; self.pr=float(self.r); self.pc=float(self.c)

    def _nxt(self): return (self.r+self.d[0])%GRID, (self.c+self.d[1])%GRID

    def step(self, occ, lights):
        nr,nc=self._nxt()
        self.pr=float(self.r); self.pc=float(self.c)
        if (nr,nc) in occ: self.w=True; return
        if (nr,nc) in INTERSECTIONS and (nr,nc) in lights:
            lt=lights[(nr,nc)]
            if not (lt.ok_h() if self.h else lt.ok_v()): self.w=True; return
        self.w=False; self.r,self.c=nr,nc


class World:
    def __init__(self):
        self.lights={}; self.tl_pos={}  # id(tl)→(cx,cy) en plot
        for cells, off in LIGHT_GROUPS:
            tl=TL(offset=off)
            rows=[r for r,c in cells]; cols=[c for r,c in cells]
            cx=(min(cols)+max(cols)+1)/2; cy=(min(rows)+max(rows)+1)/2
            self.tl_pos[id(tl)]=(cx,cy)
            for cell in cells: self.lights[cell]=tl

        taken,self.cars=[],[]
        for i in range(NV):
            v,tries=Car(i),0
            while (v.r,v.c) in taken and tries<300: v=Car(i); tries+=1
            taken.append((v.r,v.c)); self.cars.append(v)
        self.tick=0

    def step(self):
        self.tick+=1
        for lt in set(self.lights.values()): lt.step()
        occ={(v.r,v.c) for v in self.cars}
        random.shuffle(self.cars); no=set()
        for v in self.cars:
            occ.discard((v.r,v.c)); v.step(occ|no,self.lights); no.add((v.r,v.c))


# ── DIBUJO ─────────────────────────────────────────────
def _road(ax):
    ax.set_facecolor(C_RD)

    # Manzanas de tamaños variados (un parche por región)
    for r0,r1,c0,c1 in BLOCKS:
        h,w=r1-r0,c1-c0
        bs="round,pad=0.12" if (h>=4 or w>=4) else "round,pad=0.06"
        ax.add_patch(mpatches.FancyBboxPatch(
            (c0+0.06,r0+0.06),w-0.12,h-0.12,
            boxstyle=bs, facecolor=C_BLK, edgecolor="#6A8860",
            linewidth=0.5, zorder=1))

    # Línea SÓLIDA amarilla que divide los dos carriles de la avenida
    for c0,c1 in COL_GAPS:
        ax.plot([c0,c1],[AVE_ROW_DIV,AVE_ROW_DIV], color=C_DIV, lw=1.8, alpha=0.8, zorder=2)
    for r0,r1 in ROW_GAPS:
        ax.plot([AVE_COL_DIV,AVE_COL_DIV],[r0,r1], color=C_DIV, lw=1.8, alpha=0.8, zorder=2)

    # Líneas centrales punteadas (solo en tramos, no en intersecciones)
    for r in STREET_ROWS:
        for c0,c1 in COL_GAPS:
            ax.plot([c0,c1],[r+0.5,r+0.5], color="#999", lw=0.35,
                    ls=(0,(4,10)), alpha=0.22, zorder=2)
    for c in STREET_COLS:
        for r0,r1 in ROW_GAPS:
            ax.plot([c+0.5,c+0.5],[r0,r1], color="#999", lw=0.35,
                    ls=(0,(4,10)), alpha=0.22, zorder=2)


def _lv(ax, cx, ty, ph):
    """Housing vertical 3 lámparas (para tráfico H)."""
    ax.add_patch(mpatches.Rectangle(
        (cx-0.07,ty-0.27),0.14,0.30, facecolor="#0D0D0D",edgecolor="#444",lw=0.3,zorder=3))
    for i,(p,col) in enumerate([(PH_R,LRED),(PH_Y,LYEL),(PH_G,LGRN)]):
        y=ty-i*0.09; lit=(ph==p); lc=col if lit else LOFF
        if lit:
            ax.add_patch(mpatches.Circle((cx,y),0.072,color=col,alpha=0.15,zorder=3.4))
            ax.add_patch(mpatches.Circle((cx,y),0.048,color=col,alpha=0.40,zorder=3.5))
        ax.add_patch(mpatches.Circle((cx,y),0.032,color=lc,zorder=4))


def _lh(ax, lx, cy, ph):
    """Housing horizontal 3 lámparas (para tráfico V)."""
    ax.add_patch(mpatches.Rectangle(
        (lx,cy-0.07),0.26,0.14, facecolor="#0D0D0D",edgecolor="#444",lw=0.3,zorder=3))
    for i,(p,col) in enumerate([(PH_R,LRED),(PH_Y,LYEL),(PH_G,LGRN)]):
        x=lx+i*0.082; lit=(ph==p); lc=col if lit else LOFF
        if lit:
            ax.add_patch(mpatches.Circle((x,cy),0.072,color=col,alpha=0.15,zorder=3.4))
            ax.add_patch(mpatches.Circle((x,cy),0.048,color=col,alpha=0.40,zorder=3.5))
        ax.add_patch(mpatches.Circle((x,cy),0.032,color=lc,zorder=4))


def _light(ax, cx, cy, lt):
    _lv(ax, cx-0.24, cy+0.26, lt.hp())
    _lh(ax, cx+0.06, cy-0.24, lt.vp())


def _car(ax, dr, dc, v):
    cx,cy=dc+0.5, dr+0.5                      # centro exacto del carril
    face="#CC1800" if v.w else v.color
    edge="#FF6060" if v.w else "#888888"
    w,h=(0.46,0.21) if v.h else (0.21,0.46)
    ax.add_patch(mpatches.FancyBboxPatch(
        (cx-w/2,cy-h/2),w,h, boxstyle="round,pad=0.04",
        facecolor=face,edgecolor=edge,lw=0.7,zorder=5))
    ddr,ddc=v.d
    wx,wy=(cx+ddc*0.08-0.055,cy-0.055) if v.h else (cx-0.055,cy+ddr*0.08-0.055)
    ax.add_patch(mpatches.FancyBboxPatch(
        (wx,wy),0.11,0.11, boxstyle="round,pad=0.02",
        facecolor="#5BA4CF",edgecolor="none",lw=0,zorder=6,alpha=0.55))
    ax.annotate("",
        xy=(cx+ddc*0.17,cy+ddr*0.17), xytext=(cx-ddc*0.03,cy-ddr*0.03),
        arrowprops=dict(arrowstyle="->",color="white",lw=0.8),zorder=7)


# ── ANIMACIÓN ──────────────────────────────────────────
random.seed(42)
world=World()
fig,ax=plt.subplots(figsize=(8,8))
fig.patch.set_facecolor(C_BG)
_sub=[0]

def init():
    ax.clear()
    ax.set_xlim(0,GRID); ax.set_ylim(0,GRID)
    ax.set_aspect("equal"); ax.set_xticks([]); ax.set_yticks([])
    _road(ax)
    return []

def update(frame):
    sub=_sub[0]; t=sub/SUBSTEPS
    if sub==0: world.step()
    ax.clear()
    ax.set_xlim(0,GRID); ax.set_ylim(0,GRID)
    ax.set_aspect("equal"); ax.set_xticks([]); ax.set_yticks([])
    waiting=sum(1 for v in world.cars if v.w)
    ax.set_title(f"Tick {world.tick:03d}  |  Esperando: {waiting}/{NV}",
                 fontsize=10,color="#E2E8F0",pad=8)
    _road(ax)
    seen=set()
    for (_r,_c),lt in world.lights.items():
        if id(lt) not in seen:
            cx,cy=world.tl_pos[id(lt)]; _light(ax,cx,cy,lt); seen.add(id(lt))
    for v in world.cars:
        ddr=v.r-v.pr; ddc=v.c-v.pc
        ir=(float(v.r) if abs(ddr)>2 else v.pr+t*ddr)
        ic=(float(v.c) if abs(ddc)>2 else v.pc+t*ddc)
        _car(ax,ir,ic,v)
    _sub[0]=(sub+1)%SUBSTEPS
    return []

anim=FuncAnimation(fig,update,frames=TICKS*SUBSTEPS,init_func=init,blit=False,interval=60)
plt.close()
HTML(anim.to_jshtml())
```
